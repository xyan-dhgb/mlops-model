```python
# ===== CELL 3: Hàm tiền xử lý dữ liệu =====
def preprocess_csv_data(df):
    """
    Hàm tiền xử lý dữ liệu cho dataset ung thư da

    Args:
        df: DataFrame gốc

    Returns:
        df_processed: DataFrame đã được tiền xử lý
        preprocessing_report: Báo cáo quá trình tiền xử lý
    """

    print(" Bắt đầu tiền xử lý dữ liệu...")
    df_processed = df.copy()
    preprocessing_report = {}

    # 1. Thông tin ban đầu
    initial_shape = df_processed.shape
    preprocessing_report['initial_shape'] = initial_shape
    print(f" Kích thước ban đầu: {initial_shape}")

    # 2. Xử lý tên cột - chuẩn hóa
    print("\n  Chuẩn hóa tên cột...")
    old_columns = df_processed.columns.tolist()

    # Loại bỏ khoảng trắng, chuyển về lowercase
    df_processed.columns = df_processed.columns.str.strip().str.lower()

    # Thay thế khoảng trắng và ký tự đặc biệt bằng underscore
    df_processed.columns = df_processed.columns.str.replace(' ', '_')
    df_processed.columns = df_processed.columns.str.replace('[^a-zA-Z0-9_]', '_', regex=True)

    # Loại bỏ underscore thừa
    df_processed.columns = df_processed.columns.str.replace('_+', '_', regex=True)
    df_processed.columns = df_processed.columns.str.strip('_')

    new_columns = df_processed.columns.tolist()
    preprocessing_report['column_mapping'] = dict(zip(old_columns, new_columns))

    print(f"   Đã chuẩn hóa {len(old_columns)} tên cột")

    # 3. Xử lý dữ liệu thiếu
    print("\n Xử lý dữ liệu thiếu...")
    missing_before = df_processed.isnull().sum().sum()
    missing_by_column = df_processed.isnull().sum()

    if missing_before > 0:
        print(f"  Tổng dữ liệu thiếu: {missing_before}")

        # Xử lý từng cột
        for col in df_processed.columns:
            missing_count = df_processed[col].isnull().sum()
            missing_percent = (missing_count / len(df_processed)) * 100

            if missing_count > 0:
                print(f"   - {col}: {missing_count} ({missing_percent:.1f}%)")

                # Nếu thiếu quá nhiều (>70%), xem xét xóa cột
                if missing_percent > 70:
                    print(f"     Cột {col} thiếu quá nhiều dữ liệu, xem xét xóa")
                    df_processed.drop(columns=[col], inplace=True)
                    continue

                # Xử lý theo kiểu dữ liệu
                if df_processed[col].dtype in ['int64', 'float64']:
                    # Dữ liệu số: điền median
                    median_val = df_processed[col].median()
                    df_processed[col].fillna(median_val, inplace=True)
                    print(f"    Điền median ({median_val}) cho cột số")

                elif df_processed[col].dtype == 'object':
                    # Dữ liệu categorical: điền mode
                    mode_val = df_processed[col].mode()
                    if len(mode_val) > 0:
                        df_processed[col].fillna(mode_val[0], inplace=True)
                        print(f"  Điền mode ('{mode_val[0]}') cho cột categorical")
                    else:
                        df_processed[col].fillna('Unknown', inplace=True)
                        print(f"  Điền 'Unknown' cho cột categorical")
    else:
        print("   Không có dữ liệu thiếu")

    preprocessing_report['missing_before'] = missing_before
    preprocessing_report['missing_after'] = df_processed.isnull().sum().sum()

    # 4. Xử lý outliers cho dữ liệu số
    print("\n Xử lý outliers...")
    numeric_columns = df_processed.select_dtypes(include=['int64', 'float64']).columns
    outlier_info = {}

    for col in numeric_columns:
        Q1 = df_processed[col].quantile(0.25)
        Q3 = df_processed[col].quantile(0.75)
        IQR = Q3 - Q1

        # Định nghĩa outliers
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df_processed[(df_processed[col] < lower_bound) |
                               (df_processed[col] > upper_bound)]

        if len(outliers) > 0:
            outlier_count = len(outliers)
            outlier_percent = (outlier_count / len(df_processed)) * 100

            print(f"   - {col}: {outlier_count} outliers ({outlier_percent:.1f}%)")

            # Nếu outliers < 5%, có thể xóa. Nếu không, cap values
            if outlier_percent < 5:
                print(f"   Giữ nguyên outliers (< 5%)")
            else:
                # Cap outliers về boundary values
                df_processed.loc[df_processed[col] < lower_bound, col] = lower_bound
                df_processed.loc[df_processed[col] > upper_bound, col] = upper_bound
                print(f"   Đã cap outliers về [{lower_bound:.2f}, {upper_bound:.2f}]")

            outlier_info[col] = {
                'count': outlier_count,
                'percent': outlier_percent,
                'bounds': (lower_bound, upper_bound)
            }

    preprocessing_report['outliers'] = outlier_info

    # 5. Chuẩn hóa dữ liệu categorical
    print("\n  Chuẩn hóa dữ liệu categorical...")
    categorical_columns = df_processed.select_dtypes(include=['object']).columns

    for col in categorical_columns:
        # Chuẩn hóa text: loại bỏ khoảng trắng, chuyển về title case
        df_processed[col] = df_processed[col].astype(str).str.strip()
        df_processed[col] = df_processed[col].str.title()

        # Xử lý một số trường hợp đặc biệt
        df_processed[col] = df_processed[col].replace({
            'Nan': 'Unknown',
            'None': 'Unknown',
            '': 'Unknown'
        })

        unique_count = df_processed[col].nunique()
        print(f"   - {col}: {unique_count} unique values")

        # Nếu có quá nhiều unique values, có thể cần group
        if unique_count > 20:
            print(f"   Cột {col} có quá nhiều categories, xem xét group lại")

    # 6. Tạo cột mới nếu cần
    print("\n Tạo features mới...")

    # Nếu có cột age, tạo age groups
    if 'age' in df_processed.columns:
        df_processed['age_group'] = pd.cut(df_processed['age'],
                                          bins=[0, 18, 35, 50, 65, 100],
                                          labels=['<18', '18-35', '36-50', '51-65', '>65'])
        print("   Tạo cột 'age_group'")

    # Nếu có cột diameter, tạo size categories
    diameter_cols = [col for col in df_processed.columns if 'diameter' in col]
    if diameter_cols:
        # Lấy diameter trung bình nếu có nhiều cột
        if len(diameter_cols) > 1:
            df_processed['avg_diameter'] = df_processed[diameter_cols].mean(axis=1)
        else:
            df_processed['avg_diameter'] = df_processed[diameter_cols[0]]

        # Tạo size categories
        df_processed['lesion_size'] = pd.cut(df_processed['avg_diameter'],
                                           bins=[0, 5, 10, 20, float('inf')],
                                           labels=['Small', 'Medium', 'Large', 'Very Large'])
        print("   Tạo cột 'lesion_size'")

    # 7. Tổng kết
    final_shape = df_processed.shape
    preprocessing_report['final_shape'] = final_shape

    print(f"\n Tiền xử lý hoàn tất!")
    print(f" Kích thước cuối: {final_shape}")
    print(f" Thay đổi: {initial_shape[0] - final_shape[0]} dòng, {final_shape[1] - initial_shape[1]} cột")

    return df_processed, preprocessing_report
```

```python
# ===== CELL 4: Áp dụng tiền xử lý =====
print(" Áp dụng tiền xử lý cho dataset...")
df_processed, report = preprocess_csv_data(df)

# Hiển thị báo cáo tiền xử lý
print("\n BÁO CÁO TIỀN XỬ LÝ:")
print(f"Kích thước ban đầu: {report['initial_shape']}")
print(f"Kích thước sau xử lý: {report['final_shape']}")
print(f"Dữ liệu thiếu trước: {report['missing_before']}")
print(f"Dữ liệu thiếu sau: {report['missing_after']}")

if report['outliers']:
    print("\nOutliers được xử lý:")
    for col, info in report['outliers'].items():
        print(f"- {col}: {info['count']} outliers ({info['percent']:.1f}%)")

# Cập nhật df để sử dụng cho các phân tích tiếp theo
df = df_processed.copy()
print("\n Dataset đã được tiền xử lý và sẵn sàng cho phân tích!")
```

```python
# Hàm gộp các thư mục ảnh thành một thư mục

import os
import shutil
from pathlib import Path

def merge_image_folders(source_folders, destination_folder, copy_mode=True):
    """
    Gộp nhiều thư mục ảnh thành một thư mục
    Merge multiple image folders into one

    Args:
        source_folders: List các đường dẫn thư mục nguồn
        destination_folder: Đường dẫn thư mục đích
        copy_mode: True = copy file, False = move file
    """

    # Tạo thư mục đích nếu chưa tồn tại
    os.makedirs(destination_folder, exist_ok=True)

    total_files = 0
    processed_files = 0
    duplicates = 0
    errors = 0

    print(f"Bắt đầu gộp ảnh vào: {destination_folder}")
    print(f"Chế độ: {'Copy' if copy_mode else 'Move'}")
    print("-" * 50)

    for folder in source_folders:
        if not os.path.exists(folder):
            print(f"  Thư mục không tồn tại: {folder}")
            continue

        print(f" Đang xử lý: {folder}")
        folder_files = 0

        # Lấy tất cả file ảnh trong thư mục
        for file_name in os.listdir(folder):
            if file_name.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tiff', '.gif')):
                total_files += 1
                folder_files += 1

                source_path = os.path.join(folder, file_name)
                dest_path = os.path.join(destination_folder, file_name)

                try:
                    # Kiểm tra file đã tồn tại
                    if os.path.exists(dest_path):
                        # Kiểm tra xem có giống nhau không
                        if os.path.getsize(source_path) == os.path.getsize(dest_path):
                            duplicates += 1
                            print(f"    File trùng: {file_name}")
                            continue
                        else:
                            # Thêm số vào tên file
                            name, ext = os.path.splitext(file_name)
                            counter = 1
                            while os.path.exists(dest_path):
                                new_name = f"{name}_{counter}{ext}"
                                dest_path = os.path.join(destination_folder, new_name)
                                counter += 1
                            print(f"   Đổi tên: {file_name} -> {os.path.basename(dest_path)}")

                    # Copy hoặc move file
                    if copy_mode:
                        shutil.copy2(source_path, dest_path)
                    else:
                        shutil.move(source_path, dest_path)

                    processed_files += 1

                except Exception as e:
                    errors += 1
                    print(f"   Lỗi với {file_name}: {e}")

        print(f"   Hoàn thành: {folder_files} file")

    # Tóm tắt kết quả
    print("-" * 50)
    print(" KẾT QUẢ:")
    print(f"   Tổng số file tìm thấy: {total_files}")
    print(f"   File đã xử lý: {processed_files}")
    print(f"    File trùng lặp: {duplicates}")
    print(f"   Lỗi: {errors}")
    print(f"   Thư mục đích: {destination_folder}")

def merge_with_structure_check(base_path, destination_folder):
    """
    Tự động tìm và gộp các thư mục imgs_part_*
    """
    import glob

    # Tìm tất cả thư mục imgs_part_*
    part_folders = []

    # Tìm pattern imgs_part_1, imgs_part_2, etc.
    for pattern in ['imgs_part_*', 'imgs_part_*/imgs_part_*']:
        folders = glob.glob(os.path.join(base_path, pattern))
        for folder in folders:
            if os.path.isdir(folder):
                part_folders.append(folder)

    # Loại bỏ duplicate và sort
    part_folders = sorted(list(set(part_folders)))

    if not part_folders:
        print(" Không tìm thấy thư mục imgs_part_* nào!")
        return

    print(f" Tìm thấy {len(part_folders)} thư mục:")
    for folder in part_folders:
        file_count = len([f for f in os.listdir(folder)
                         if f.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tiff', '.gif'))])
        print(f"   {folder} ({file_count} ảnh)")

    # Gộp tất cả
    merge_image_folders(part_folders, destination_folder, copy_mode=True)
```

```python
# ===== CELL 3: Hàm tải và xử lý hình ảnh (Đã sửa lỗi) =====

# Thêm các import cần thiết
import os
import numpy as np
import cv2
from PIL import Image, ImageEnhance
from sklearn.preprocessing import LabelEncoder

def load_image(image_path, target_size=(224, 224), color_mode='rgb'):
    """
    Tải và resize hình ảnh
    install and resize images

    Args:
        image_path: Đường dẫn đến hình ảnh
        target_size: Kích thước mục tiêu (width, height)
        color_mode: 'rgb', 'grayscale', hoặc 'rgba'

    Returns:
        numpy array của hình ảnh đã được xử lý
    """
    try:
        # Mở hình ảnh bằng PIL
        img = Image.open(image_path)

        # Chuyển đổi color mode
        if color_mode.lower() == 'rgb':
            img = img.convert('RGB')
        elif color_mode.lower() == 'grayscale':
            img = img.convert('L')
        elif color_mode.lower() == 'rgba':
            img = img.convert('RGBA')

        # Resize hình ảnh
        img = img.resize(target_size, Image.Resampling.LANCZOS)

        # Chuyển thành numpy array
        img_array = np.array(img)

        # Normalize pixel values về [0, 1]
        if img_array.dtype == np.uint8:
            img_array = img_array.astype(np.float32) / 255.0

        return img_array

    except Exception as e:
        print(f"Lỗi khi tải ảnh {image_path}: {e}")
        return None

def preprocess_image(img_array, apply_clahe=True, apply_gaussian=True, enhance_contrast=1.2):
    """
    Tiền xử lý hình ảnh cho phân tích đa
    preprocess images for multi-modal analysis
    Args:
        img_array: Numpy array của hình ảnh (đã normalize)
        apply_clahe: Có áp dụng CLAHE (Contrast Limited Adaptive Histogram Equalization)
        apply_gaussian: Có áp dụng Gaussian blur để giảm noise
        enhance_contrast: Hệ số tăng cường contrast

    Returns:
        Hình ảnh đã được tiền xử lý
    """
    if img_array is None:
        return None

    try:
        # Chuyển về uint8 để xử lý với OpenCV
        if img_array.dtype == np.float32:
            img_uint8 = (img_array * 255).astype(np.uint8)
        else:
            img_uint8 = img_array.copy()

        processed_img = img_uint8.copy()

        # Áp dụng CLAHE để cải thiện contrast local
        if apply_clahe and len(processed_img.shape) == 3:
            # Chuyển sang LAB color space
            lab = cv2.cvtColor(processed_img, cv2.COLOR_RGB2LAB)
            l, a, b = cv2.split(lab)

            # Áp dụng CLAHE cho kênh L (lightness)
            clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
            l = clahe.apply(l)

            # Ghép lại và chuyển về RGB
            lab = cv2.merge([l, a, b])
            processed_img = cv2.cvtColor(lab, cv2.COLOR_LAB2RGB)

        # Áp dụng Gaussian blur nhẹ để giảm noise
        if apply_gaussian:
            processed_img = cv2.GaussianBlur(processed_img, (3, 3), 0)

        # Tăng cường contrast
        if enhance_contrast != 1.0:
            processed_img = cv2.convertScaleAbs(processed_img, alpha=enhance_contrast, beta=0)

        # Normalize về [0, 1]
        processed_img = processed_img.astype(np.float32) / 255.0

        return processed_img

    except Exception as e:
        print(f"Lỗi trong tiền xử lý: {e}")
        return img_array

def augment_image(img_array, rotation_range=15, brightness_range=(0.8, 1.2), zoom_range=0.1):
    """
    Data augmentation cho hình ảnh

    Args:
        img_array: Numpy array của hình ảnh
        rotation_range: Độ xoay tối đa (degrees)
        brightness_range: Khoảng thay đổi độ sáng
        zoom_range: Khoảng zoom in/out

    Returns:
        Hình ảnh đã được augment
    """
    if img_array is None:
        return None

    try:
        # Chuyển về PIL Image
        if img_array.dtype == np.float32:
            pil_img = Image.fromarray((img_array * 255).astype(np.uint8))
        else:
            pil_img = Image.fromarray(img_array)

        # Random rotation
        if rotation_range > 0:
            angle = np.random.uniform(-rotation_range, rotation_range)
            pil_img = pil_img.rotate(angle, fillcolor='white')

        # Random brightness
        if brightness_range != (1.0, 1.0):
            brightness_factor = np.random.uniform(brightness_range[0], brightness_range[1])
            enhancer = ImageEnhance.Brightness(pil_img)
            pil_img = enhancer.enhance(brightness_factor)

        # Random zoom (thực hiện bằng cách resize và crop)
        if zoom_range > 0:
            zoom_factor = np.random.uniform(1 - zoom_range, 1 + zoom_range)
            w, h = pil_img.size
            new_w, new_h = int(w * zoom_factor), int(h * zoom_factor)
            pil_img = pil_img.resize((new_w, new_h), Image.Resampling.LANCZOS)

            # Crop về kích thước ban đầu
            if zoom_factor > 1:
                left = (new_w - w) // 2
                top = (new_h - h) // 2
                pil_img = pil_img.crop((left, top, left + w, top + h))
            else:
                # Pad if zoomed out
                pil_img = pil_img.resize((w, h), Image.Resampling.LANCZOS)

        # Chuyển về numpy array
        aug_array = np.array(pil_img).astype(np.float32) / 255.0
        return aug_array

    except Exception as e:
        print(f"Lỗi trong augmentation: {e}")
        return img_array


print("Các hàm xử lý hình ảnh đã được sửa lỗi và định nghĩa lại!")
```

```python
import matplotlib.pyplot as plt

def show_image_processing_pipeline(image_path):
    """
    Hiển thị hình ảnh gốc, sau tiền xử lý, và sau augmentation

    Args:
        image_path: đường dẫn hình ảnh cần xử lý
    """
    # Tải ảnh gốc
    img_original = load_image(image_path, target_size=(224, 224), color_mode='rgb')
    if img_original is None:
        print("Không thể tải ảnh.")
        return

    # Tiền xử lý
    img_preprocessed = preprocess_image(img_original, apply_clahe=True, apply_gaussian=True, enhance_contrast=1.2)

    # Augmentation
    img_augmented = augment_image(img_preprocessed, rotation_range=15, brightness_range=(0.8, 1.2), zoom_range=0.1)

    # Vẽ kết quả
    fig, axs = plt.subplots(1, 3, figsize=(15, 5))
    axs[0].imshow(img_original)
    axs[0].set_title("Ảnh gốc")
    axs[0].axis('off')

    axs[1].imshow(img_preprocessed)
    axs[1].set_title("Sau tiền xử lý")
    axs[1].axis('off')

    axs[2].imshow(img_augmented)
    axs[2].set_title("Sau Augmentation")
    axs[2].axis('off')

    plt.tight_layout()
    plt.show()

# === Ví dụ sử dụng ===
sample_image_path = "/content/all_images/PAT_100_393_595.png"   # Đường dẫn ảnh ví dụ
show_image_processing_pipeline(sample_image_path)
```

```python
# ===== CELL 5: Phân tích kích thước và thống kê hình ảnh =====
def analyze_image_properties(image_dir, image_files, sample_size=100):
    """
    Phân tích các thuộc tính của hình ảnh trong dataset
    Image analysis properties
    """
    print("Phân tích thuộc tính hình ảnh...")

    if len(image_files) == 0:
        print("Không có hình ảnh để phân tích!")
        return

    # Lấy mẫu ngẫu nhiên nếu dataset quá lớn
    sample_files = np.random.choice(image_files,
                                   size=min(sample_size, len(image_files)),
                                   replace=False)

    widths, heights, channels, file_sizes = [], [], [], []
    formats = []

    for img_file in sample_files:
        try:
            img_path = os.path.join(image_dir, img_file)

            # Lấy thông tin file
            file_size = os.path.getsize(img_path) / 1024  # KB
            file_sizes.append(file_size)

            # Lấy thông tin ảnh
            with Image.open(img_path) as img:
                w, h = img.size
                widths.append(w)
                heights.append(h)

                # Số kênh màu
                if img.mode == 'RGB':
                    channels.append(3)
                elif img.mode == 'RGBA':
                    channels.append(4)
                elif img.mode == 'L':
                    channels.append(1)
                else:
                    channels.append(len(img.getbands()))

                # Format
                formats.append(img.format)

        except Exception as e:
            print(f"Lỗi khi đọc {img_file}: {e}")

    # Tạo DataFrame thống kê
    stats_df = pd.DataFrame({
        'width': widths,
        'height': heights,
        'channels': channels,
        'file_size_kb': file_sizes,
        'format': formats
    })

    # Hiển thị thống kê
    print(f"\nTHỐNG KÊ HÌNH ẢNH (mẫu {len(sample_files)} ảnh):")
    print(f"Kích thước:")
    print(f"   - Width: {np.mean(widths):.0f} ± {np.std(widths):.0f} pixels")
    print(f"   - Height: {np.mean(heights):.0f} ± {np.std(heights):.0f} pixels")
    print(f"   - Min size: {np.min(widths)}x{np.min(heights)}")
    print(f"   - Max size: {np.max(widths)}x{np.max(heights)}")

    print(f"\nChannels: {dict(pd.Series(channels).value_counts())}")
    print(f"File size: {np.mean(file_sizes):.1f} ± {np.std(file_sizes):.1f} KB")
    print(f"Formats: {dict(pd.Series(formats).value_counts())}")

    # Vẽ biểu đồ phân tích
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))

    # Phân bố kích thước
    axes[0, 0].scatter(widths, heights, alpha=0.6, color='blue')
    axes[0, 0].set_xlabel('Width (pixels)')
    axes[0, 0].set_ylabel('Height (pixels)')
    axes[0, 0].set_title('Phân bố kích thước ảnh')
    axes[0, 0].grid(True, alpha=0.3)

    # Histogram width
    axes[0, 1].hist(widths, bins=20, color='skyblue', alpha=0.7, edgecolor='black')
    axes[0, 1].set_xlabel('Width (pixels)')
    axes[0, 1].set_ylabel('Frequency')
    axes[0, 1].set_title('Phân bố Width')
    axes[0, 1].grid(True, alpha=0.3)

    # Histogram height
    axes[1, 0].hist(heights, bins=20, color='lightcoral', alpha=0.7, edgecolor='black')
    axes[1, 0].set_xlabel('Height (pixels)')
    axes[1, 0].set_ylabel('Frequency')
    axes[1, 0].set_title('Phân bố Height')
    axes[1, 0].grid(True, alpha=0.3)

    # File size distribution
    axes[1, 1].hist(file_sizes, bins=20, color='lightgreen', alpha=0.7, edgecolor='black')
    axes[1, 1].set_xlabel('File Size (KB)')
    axes[1, 1].set_ylabel('Frequency')
    axes[1, 1].set_title('Phân bố kích thước file')
    axes[1, 1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    return stats_df

# Chạy phân tích
if image_files:
    image_stats = analyze_image_properties(image_dir, image_files)
```

```python
def prepare_multimodal_data(df, image_dir, target_size=(224, 224), is_training=True):
    """
    Chuẩn bị dữ liệu multimodal (ảnh + bảng) với xử lý lỗi tốt hơn
    prepare multimodal data with better error handling
    """
    X_tabular, X_image, y = [], [], []

    # Encode nhãn
    label_encoder = LabelEncoder()
    y_encoded = label_encoder.fit_transform(df["diagnostic"])

    # Xác định các cột dùng cho dữ liệu bảng
    numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
    tabular_cols = [col for col in numeric_cols if col not in ['diagnostic', 'img_id']]

    print(f"Số đặc trưng bảng: {len(tabular_cols)}")
    print(f"Các cột bảng: {tabular_cols}")

    # Xử lý missing values trong dữ liệu bảng
    df_clean = df.copy()

    # Impute missing values cho các cột số
    from sklearn.impute import SimpleImputer
    imputer = SimpleImputer(strategy='median')  # hoặc 'mean'

    if tabular_cols:
        df_clean[tabular_cols] = imputer.fit_transform(df_clean[tabular_cols])
        print(f"Đã xử lý missing values cho {len(tabular_cols)} cột")

    successful_count = 0
    total_count = len(df_clean)

    for idx, row in df_clean.iterrows():
        try:
            # Xử lý tên file ảnh - sửa lỗi chính ở đây
            image_name = str(row["img_id"]).strip()

            # Kiểm tra xem có phải định dạng đúng không
            if not image_name.lower().endswith(('.jpg', '.jpeg', '.png')):
                # Thêm extension .png nếu chưa có
                image_name += '.png'

            # Tạo đường dẫn đầy đủ
            image_path = os.path.join(image_dir, image_name)

            # Kiểm tra file có tồn tại không
            if not os.path.exists(image_path):
                # Thử tìm file với các extension khác
                base_name = os.path.splitext(image_name)[0]
                found = False

                for ext in ['.png', '.jpg', '.jpeg']:
                    alt_path = os.path.join(image_dir, base_name + ext)
                    if os.path.exists(alt_path):
                        image_path = alt_path
                        found = True
                        break

                if not found:
                    print(f"Không tìm thấy ảnh: {image_name} (bỏ qua dòng {idx})")
                    continue

            # Load ảnh
            img = load_image(image_path, target_size=target_size)
            if img is None:
                print(f"Lỗi khi load ảnh: {image_path} (bỏ qua dòng {idx})")
                continue

            # Tiền xử lý và augmentation
            img = preprocess_image(img)
            if img is None:
                print(f"Lỗi khi tiền xử lý ảnh: {image_path} (bỏ qua dòng {idx})")
                continue

            if is_training:
                img = augment_image(img)
                if img is None:
                    print(f"Lỗi khi augment ảnh: {image_path} (bỏ qua dòng {idx})")
                    continue

            # Dữ liệu bảng - đã được xử lý missing values
            if tabular_cols:
                tabular_features = row[tabular_cols].values.astype(np.float32)

                # Kiểm tra NaN values (không nên có sau khi impute)
                if np.isnan(tabular_features).any():
                    print(f"Vẫn còn NaN sau imputation ở dòng {idx} (bỏ qua)")
                    continue
            else:
                # Nếu không có cột tabular, tạo vector rỗng
                tabular_features = np.array([], dtype=np.float32)

            # Nếu tất cả đều OK, thêm vào danh sách
            X_image.append(img)
            X_tabular.append(tabular_features)
            y.append(y_encoded[idx])
            successful_count += 1

        except Exception as e:
            print(f"Lỗi không mong đợi ở dòng {idx}: {e} (bỏ qua)")
            continue

    if not X_tabular or not X_image:
        print("Không có dữ liệu hợp lệ sau khi xử lý.")
        return np.array([]), np.array([]), np.array([]), label_encoder

    print(f"Đã xử lý thành công {successful_count}/{total_count} mẫu dữ liệu")

    # Xử lý trường hợp tabular_cols rỗng
    if tabular_cols:
        X_tabular_array = np.array(X_tabular, dtype=np.float32)
    else:
        X_tabular_array = np.empty((len(X_image), 0), dtype=np.float32)

    return (
        X_tabular_array,
        np.array(X_image, dtype=np.float32),
        np.array(y),
        label_encoder
    )
```

```python
# Cell: Xây dựng mô hình đa phương thức (multimodal)
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Dense, Dropout, Flatten, Concatenate, Conv2D, MaxPooling2D, BatchNormalization, GlobalAveragePooling2D
)

def build_multimodal_model(tabular_shape, image_shape, num_classes=3):
    """
    Xây dựng mô hình kết hợp dữ liệu hình ảnh và dữ liệu bảng.
    Build a multimodal model combining image and tabular data.

    Parameters:
    - tabular_shape: Kích thước đầu vào của dữ liệu bảng (ví dụ: (6,))
    - image_shape: Kích thước đầu vào của ảnh (ví dụ: (224, 224, 3))
    - num_classes: Số lớp phân loại.

    Returns:
    - model: Mô hình Keras đã sẵn sàng để huấn luyện.
    """

    # Nhánh xử lý ảnh (CNN đơn giản hoặc có thể dùng base model như EfficientNet)
    image_input = Input(shape=image_shape, name='image_input')
    x = Conv2D(32, (3, 3), activation='relu')(image_input)
    x = MaxPooling2D((2, 2))(x)
    x = BatchNormalization()(x)

    x = Conv2D(64, (3, 3), activation='relu')(x)
    x = MaxPooling2D((2, 2))(x)
    x = BatchNormalization()(x)

    x = GlobalAveragePooling2D()(x)
    x = Dense(128, activation='relu')(x)
    x = Dropout(0.3)(x)

    # Nhánh xử lý dữ liệu bảng
    tabular_input = Input(shape=tabular_shape, name='tabular_input')
    y = Dense(64, activation='relu')(tabular_input)
    y = Dropout(0.3)(y)
    y = Dense(32, activation='relu')(y)

    # Kết hợp hai nhánh
    combined = Concatenate()([x, y])
    z = Dense(64, activation='relu')(combined)
    z = Dropout(0.3)(z)
    output = Dense(num_classes, activation='softmax')(z)

    # Tạo mô hình
    model = Model(inputs=[image_input, tabular_input], outputs=output)

    # Compile mô hình
    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

    return model
```

```python
# Cell: Huấn luyện mô hình đa phương thức
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

def train_model(model,
                X_tab_train, X_img_train, y_train,
                X_tab_val, X_img_val, y_val,
                epochs=20, batch_size=32):
    """
    Huấn luyện mô hình đa phương thức với dữ liệu huấn luyện và validation.
    Train a multimodal model with training and validation data.
    Parameters:
    - model: Mô hình Keras đã được xây dựng.
    - X_tab_train: Dữ liệu bảng cho tập huấn luyện.
    - X_img_train: Dữ liệu ảnh cho tập huấn luyện.
    - y_train: Nhãn tập huấn luyện.
    - X_tab_val: Dữ liệu bảng cho tập validation.
    - X_img_val: Dữ liệu ảnh cho tập validation.
    - y_val: Nhãn tập validation.
    - epochs: Số vòng lặp huấn luyện (default: 20).
    - batch_size: Kích thước batch (default: 32).

    Returns:
    - history: Đối tượng history sau khi huấn luyện.
    """

    # Callbacks để dừng sớm nếu quá khớp và lưu mô hình tốt nhất
    early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
    checkpoint = ModelCheckpoint('best_multimodal_model.h5',
                                 monitor='val_loss',
                                 save_best_only=True,
                                 verbose=1)

    history = model.fit(
        {'image_input': X_img_train, 'tabular_input': X_tab_train},
        y_train,
        validation_data=({'image_input': X_img_val, 'tabular_input': X_tab_val}, y_val),
        epochs=epochs,
        batch_size=batch_size,
        callbacks=[early_stop, checkpoint],
        verbose=1
    )

    return history
```

```python
# Cell: Vẽ biểu đồ accuracy và loss
import matplotlib.pyplot as plt

def plot_training_history(history):
    """
    Vẽ biểu đồ accuracy và loss trong quá trình huấn luyện.
    Draw accuracy and loss during training.
    Parameters:
    - history: Đối tượng history trả về từ model.fit().
    """
    acc = history.history['accuracy']
    val_acc = history.history['val_accuracy']
    loss = history.history['loss']
    val_loss = history.history['val_loss']
    epochs = range(1, len(acc) + 1)

    # Biểu đồ Accuracy
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.plot(epochs, acc, 'bo-', label='Training Accuracy')
    plt.plot(epochs, val_acc, 'ro-', label='Validation Accuracy')
    plt.title('Training and Validation Accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend()

    # Biểu đồ Loss
    plt.subplot(1, 2, 2)
    plt.plot(epochs, loss, 'bo-', label='Training Loss')
    plt.plot(epochs, val_loss, 'ro-', label='Validation Loss')
    plt.title('Training and Validation Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()

    plt.tight_layout()
    plt.show()
```

```python
# Cell: Đánh giá mô hình trên tập test với ROC curves
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, roc_curve, auc
from sklearn.preprocessing import label_binarize
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from itertools import cycle

def evaluate_model(model, X_tab_test, X_img_test, y_test, label_encoder=None):
    """
    Đánh giá mô hình đa phương thức trên tập test và hiển thị kết quả bao gồm ROC curves.
    Evaluate a multimodal model on the test set and display ROC curves.
    Parameters:
    - model: Mô hình Keras đã huấn luyện. (Model keras trained)
    - X_tab_test: Dữ liệu bảng cho tập test. (Table data for test file)
    - X_img_test: Dữ liệu ảnh cho tập test. (Image data for test file)
    - y_test: Nhãn thực tế cho tập test. (True labels for test file)
    - label_encoder: (tùy chọn) LabelEncoder dùng để hiển thị tên lớp thay vì số.

    Returns:
    - accuracy: Độ chính xác của mô hình. (Accuracy of model)
    """
    # Dự đoán
    y_pred_probs = model.predict({'image_input': X_img_test, 'tabular_input': X_tab_test})
    y_pred = np.argmax(y_pred_probs, axis=1)

    # Độ chính xác
    acc = accuracy_score(y_test, y_pred)
    print(f"Accuracy: {acc:.4f}")

    # Báo cáo phân loại
    target_names = label_encoder.classes_ if label_encoder else [str(i) for i in np.unique(y_test)]
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, target_names=target_names))

    # Ma trận nhầm lẫn
    plt.figure(figsize=(8, 6))
    cm = confusion_matrix(y_test, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=target_names, yticklabels=target_names)
    plt.xlabel('Predicted')
    plt.ylabel('True')
    plt.title('Confusion Matrix')
    plt.tight_layout()
    plt.show()

    # ROC Curves
    n_classes = len(target_names)

    if n_classes == 2:
        # Binary classification
        fpr, tpr, _ = roc_curve(y_test, y_pred_probs[:, 1])
        roc_auc = auc(fpr, tpr)

        plt.figure(figsize=(8, 6))
        plt.plot(fpr, tpr, color='darkorange', lw=2,
                label=f'ROC curve (AUC = {roc_auc:.3f})')
        plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
        plt.xlim([0.0, 1.0])
        plt.ylim([0.0, 1.05])
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title('Receiver Operating Characteristic (ROC) Curve')
        plt.legend(loc="lower right")
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()

    else:
        # Multi-class classification
        # Binarize the output
        y_test_bin = label_binarize(y_test, classes=range(n_classes))

        # Compute ROC curve and ROC area for each class
        fpr = dict()
        tpr = dict()
        roc_auc = dict()

        for i in range(n_classes):
            fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], y_pred_probs[:, i])
            roc_auc[i] = auc(fpr[i], tpr[i])

        # Compute micro-average ROC curve and ROC area
        fpr["micro"], tpr["micro"], _ = roc_curve(y_test_bin.ravel(), y_pred_probs.ravel())
        roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])

        # Compute macro-average ROC curve and ROC area
        # First aggregate all false positive rates
        all_fpr = np.unique(np.concatenate([fpr[i] for i in range(n_classes)]))

        # Then interpolate all ROC curves at this points
        mean_tpr = np.zeros_like(all_fpr)
        for i in range(n_classes):
            mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])

        # Finally average it and compute AUC
        mean_tpr /= n_classes

        fpr["macro"] = all_fpr
        tpr["macro"] = mean_tpr
        roc_auc["macro"] = auc(fpr["macro"], tpr["macro"])

        # Plot all ROC curves
        plt.figure(figsize=(10, 8))

        # Plot micro-average ROC curve
        plt.plot(fpr["micro"], tpr["micro"],
                label=f'Micro-average (AUC = {roc_auc["micro"]:.3f})',
                color='deeppink', linestyle=':', linewidth=4)

        # Plot macro-average ROC curve
        plt.plot(fpr["macro"], tpr["macro"],
                label=f'Macro-average (AUC = {roc_auc["macro"]:.3f})',
                color='navy', linestyle=':', linewidth=4)

        # Plot ROC curves for each class
        colors = cycle(['aqua', 'darkorange', 'cornflowerblue', 'green', 'red', 'purple'])
        for i, color in zip(range(n_classes), colors):
            plt.plot(fpr[i], tpr[i], color=color, lw=2,
                    label=f'{target_names[i]} (AUC = {roc_auc[i]:.3f})')

        plt.plot([0, 1], [0, 1], 'k--', lw=2)
        plt.xlim([0.0, 1.0])
        plt.ylim([0.0, 1.05])
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title('Multi-class ROC Curves')
        plt.legend(loc="lower right")
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()

        # Print AUC scores for each class
        print("\nROC AUC Scores:")
        print(f"Micro-average: {roc_auc['micro']:.4f}")
        print(f"Macro-average: {roc_auc['macro']:.4f}")
        for i in range(n_classes):
            print(f"{target_names[i]}: {roc_auc[i]:.4f}")

    return acc
```

```python
def predict_cancer_diagnosis(model, csv_data, image_path, tabular_columns, device='cpu', encoders=None):
    import pandas as pd
    import numpy as np  # Import numpy
    import tensorflow as tf # Import tensorflow
    from PIL import Image

    # Load DataFrame
    if isinstance(csv_data, str):
        df = pd.read_csv(csv_data)
    else:
        df = csv_data

    # Lấy mẫu đầu tiên
    sample = df.iloc[0][tabular_columns].copy()

    # Tiền xử lý boolean: "TRUE"/"FALSE" thành 1/0
    bool_map = {"TRUE": 1, "FALSE": 0, "True": 1, "False": 0}
    for col in sample.index:
        val = sample[col]
        if isinstance(val, str) and val.strip().upper() in bool_map:
            sample[col] = bool_map[val.strip().upper()]

    # Mã hóa cột phân loại
    if encoders:
        for col in encoders:
            if col in sample and not pd.isna(sample[col]):
                 # Ensure input to transform is a list or array-like
                sample[col] = encoders[col].transform([str(sample[col])])[0]


    # Điền NaN còn lại bằng 0
    sample = sample.fillna(0)

    # Ép kiểu float
    try:
        sample = sample.astype('float32')
    except Exception as e:
        print("[LỖI] Không thể ép kiểu float:", e)
        print("Giá trị hiện tại:\n", sample)
        return None

    # Chuyển về TensorFlow tensor
    # Need to reshape to (1, num_features)
    tabular_input = tf.constant(sample.values, dtype=tf.float32)[tf.newaxis, ...]

    # Load ảnh và tiền xử lý (sử dụng hàm load_image đã định nghĩa)
    # Assumes load_image returns a numpy array (H, W, C) normalized [0, 1]
    img_array = load_image(image_path, target_size=(224, 224))
    if img_array is None:
        print(f"[LỖI] Không thể tải hoặc xử lý ảnh: {image_path}")
        return None

    # Apply preprocess_image if needed (assuming preprocess_image works on numpy arrays)
    img_processed = preprocess_image(img_array)
    if img_processed is None:
         print(f"[LỖI] Lỗi tiền xử lý ảnh: {image_path}")
         return None

    # Convert processed image numpy array to TensorFlow tensor and reshape
    # Need to reshape to (1, H, W, C)
    image_input = tf.constant(img_processed, dtype=tf.float32)[tf.newaxis, ...]

    # Dự đoán
    # TensorFlow models don't have .eval(), just call the model
    # Use model.predict() for inference
    y_pred_probs = model.predict({'image_input': image_input, 'tabular_input': tabular_input})

    # Decode the prediction
    # The model outputs probabilities for each class
    predicted_class_index = np.argmax(y_pred_probs, axis=1)[0]
    confidence = np.max(y_pred_probs) * 100 # Confidence is the max probability

    # Get class name if label_encoder is provided
    if encoders and 'diagnostic' in encoders: # Assuming label_encoder for diagnostic is stored under 'diagnostic' key
        predicted_diagnosis = encoders['diagnostic'].inverse_transform([predicted_class_index])[0]
    elif 'diagnostic' in globals() and isinstance(diagnostic, LabelEncoder): # Check global label_encoder
         predicted_diagnosis = diagnostic.inverse_transform([predicted_class_index])[0]
    elif 'label_encoder' in globals() and isinstance(label_encoder, LabelEncoder): # Check global label_encoder
        predicted_diagnosis = label_encoder.inverse_transform([predicted_class_index])[0]
    else:
        # Fallback to index string if no encoder is found
        predicted_diagnosis = f"Class {predicted_class_index}"
        print("[CẢNH BÁO] Không tìm thấy LabelEncoder cho 'diagnostic'. Dự đoán dựa trên chỉ số lớp.")

    # Return full probabilities
    probabilities = {
         label_encoder.classes_[i]: prob * 100
         for i, prob in enumerate(y_pred_probs[0])
     }


    return {
        'diagnosis': predicted_diagnosis,
        'confidence': confidence,
        'probabilities': probabilities,
        'predicted_class_index': predicted_class_index
    }
```

```python
# Trực quan hóa bằng hình ảnh thực tế
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

def visualize_prediction(image_path, prediction_result, heatmap=None):
    """
    Hiển thị ảnh quang của bệnh nhân kèm theo dự đoán và tùy chọn heatmap.

    Args:
        image_path (str): Đường dẫn ảnh quang.
        prediction_result (dict): Kết quả từ hàm predict_cancer_diagnosis.
        heatmap (ndarray): Mảng 2D heatmap để mô phỏng vùng nghi ngờ (giá trị 0-1). Tùy chọn.
    """
    # Load ảnh gốc
    image = Image.open(image_path).convert("RGB")
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.imshow(image)
    ax.axis('off')

    # Hiển thị chẩn đoán
    diagnosis = prediction_result['diagnosis']
    confidence = prediction_result['confidence']
    color = 'green' if diagnosis == 'Benign' else 'red'
    ax.set_title(f"{diagnosis} ({confidence:.2f}%)", fontsize=16, color=color)

    # Nếu có heatmap
    if heatmap is not None:
        heatmap_resized = np.array(Image.fromarray((heatmap * 255).astype(np.uint8)).resize(image.size))
        ax.imshow(heatmap_resized, cmap='jet', alpha=0.4)

    plt.tight_layout()
    plt.show()
```
